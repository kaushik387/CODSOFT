# Goal 
# Build an AI agent that plays Tic-tac-toe game against human 

```python
board = [" " for _ in range(9)]
print(board)
```

# Defining game functions

```python
# check winner
def check_winner(board):
    win_positions = [
        [0,1,2],[3,4,5],[6,7,8],   #rows
        [0,3,6],[1,4,7],[2,5,8],   #columns
        [0,4,8],[2,4,6]            #diagonals
    ]
    for pos in win_positions:
        if board[pos[0]] == board[pos[1]] == board[pos[2]] != " ":
            return board[pos[0]]
    return None
```

```python
# check draw
def is_draw(board):
    return " " not in board
```

# Minimax Algorithm (Core AI)

 Idea:

Try all possible moves
Assign score:
+1 → AI wins
-1 → Human wins
0 → Draw
Choose best move recursively

This explores all possibilities and selects the optimal move

```python
# Minimax code
def minimax(board, is_maximizing):
    winner = check_winner(board)

    if winner == "O":   # AI
        return 1
    elif winner == "X": # Human
        return -1
    elif is_draw(board):
        return 0

    if is_maximizing:
        best_score = -float("inf")
        for i in range(9):
            if board[i] == " ":
                board[i] = "O"
                score = minimax(board, False)
                board[i] = " "
                best_score = max(score, best_score)
        return best_score
    else:
        best_score = float("inf")
        for i in range(9):
            if board[i] == " ":
                board[i] = "X"
                score = minimax(board, True)
                board[i] = " "
                best_score = min(score, best_score)
        return best_score
```

# Best move function

```python
def best_move(board):
    best_score = -float("inf")
    move = -1

    for i in range(9):
        if board[i] == " ":
            board[i] = "O"
            score = minimax(board, False)
            board[i] = " "

            if score > best_score:
                best_score = score
                move = i

    return move
```

# Alpha- Beta pruning

```python
def minimax_ab(board, depth, alpha, beta, is_maximizing):
    winner = check_winner(board)

    if winner == "O":
        return 1
    elif winner == "X":
        return -1
    elif is_draw(board):
        return 0

    if is_maximizing:
        max_eval = -float("inf")
        for i in range(9):
            if board[i] == " ":
                board[i] = "O"
                eval = minimax_ab(board, depth+1, alpha, beta, False)
                board[i] = " "
                max_eval = max(max_eval, eval)
                alpha = max(alpha, eval)
                if beta <= alpha:
                    break   # PRUNE
        return max_eval
    else:
        min_eval = float("inf")
        for i in range(9):
            if board[i] == " ":
                board[i] = "X"
                eval = minimax_ab(board, depth+1, alpha, beta, True)
                board[i] = " "
                min_eval = min(min_eval, eval)
                beta = min(beta, eval)
                if beta <= alpha:
                    break   # PRUNE
        return min_eval
```

# Board Display


```python
def print_board(board):
    print()
    print(board[0], "|", board[1], "|", board[2])
    print("---------")
    print(board[3], "|", board[4], "|", board[5])
    print("---------")
    print(board[6], "|", board[7], "|", board[8])
    print()   
```

# Game loop (Human vs AI)

```python
while True:
    print_board(board)

    # Human move
    pos = int(input("Enter position (0-8): "))
    board[pos] = "X"

    winner = check_winner(board)
    if winner:
        print_board(board)
        if winner == "X":
            print("Human wins!")
        else:
            print("AI wins!")
        break

    if is_draw(board):
        print_board(board)
        print("Draw!")
        break

    # AI move
    ai_move = best_move(board)
    board[ai_move] = "O"

    print("AI chose:", ai_move)

    winner = check_winner(board)
    if winner:
        print_board(board)
        if winner == "X":
            print("Human wins!")
        else:
            print("AI wins!")
        break

    if is_draw(board):
        print_board(board)
        print("Draw!")
        break
```
